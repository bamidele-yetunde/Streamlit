### Data Cleaning

```python
# TS column
df['ts']  = pd. to_datetime(df['ts'])
```

```python
df.info()
```

```python
df['year'] = df['ts'].dt.year
```

```python
df['year']
```

```python
df["ts"] = pd.to_datetime(df["ts"])
df["month"] = df["ts"].dt.month
df["month_name"] = df["ts"].dt.month_name()
df["quarter"] = df["ts"].dt.quarter
df["day"] = df["ts"].dt.day
df["hour"] = df["ts"].dt.hour
df["second"] = df["ts"].dt.second
```

```python
top_art_alb =  df.groupby('artist_name') ['album_name'].nunique().sort_values(ascending=False).head(10)
top_art_alb
```

```python
top_alb_track =  df.groupby('album_name') ['track_name'].nunique().sort_values(ascending=False).head(10)
top_alb_track
```
