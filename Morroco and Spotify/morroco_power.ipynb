### Findings: Seasonal Patterns

**Hourly:** Electricity use is lowest early in the morning, around 6 AM, when most people are asleep. 
    It rises through the day and peaks around 8 PM, when people are home using lights, cooking and running appliances.

**Weekly:** Sunday has the lowest electricity use, likely because many businesses and offices are closed. 
        Weekdays are all fairly similar and slightly higher.

**Annual:** Electricity use is highest in July and August. This is most likely because 
of air conditioning during the hot summer months. December has the lowest use.

**Conclusion:** Electricity consumption follows clear patterns. 
    It depends on the time of day, is lower on weekends, and goes up a lot in summer.


### Findings: Weather and Electricity Use

Correlation measures how strongly two things are related, on a scale from -1 to 1. A value close to 1 means they rise together. A value close to -1 means one rises while the other falls. A value close to 0 means there is little or no relationship.

- **Temperature (0.49):** Moderate positive relationship. When it gets hotter, electricity use goes up.
- **Humidity (-0.30):** Weak negative relationship. When humidity is higher, electricity use goes down slightly.
- **Wind Speed (0.22):** Weak positive relationship. It has only a small effect.

**Conclusion:** Temperature has the strongest relationship with electricity consumption. This matches the summer peak we saw in Q1, since hot weather leads to more cooling.


### Findings: Zone Differences

**Share of total use:** Zone 1 uses the most electricity (about 45%), followed by Zone 2 (about 30%) and Zone 3 (about 25%).

**Daily pattern:** All three zones follow the same daily pattern. They are lowest in the early morning and peak around 8 PM.

**Yearly pattern:** The zones behave differently across the year:
- **Zone 1** stays fairly steady all year.
- **Zone 2** rises toward the end of the year, in November and December.
- **Zone 3** has the biggest change. It jumps sharply in summer (July) and drops a lot in winter.

**Conclusion:** Yes, the zones have different patterns. Their daily habits are the same, but their seasonal patterns differ. This may be because each zone has different types of users, such as homes, businesses or industry.
